## Functions

We've actually been dealing with functions all along. __Functions__ are reusable blocks of code that perform specific tasks. They help organize code, reduce repetitive code writing, and make programs easier to read and maintain. 

Things like `print()` are a function. The `()` parentheses signal that you're working with a function. The Python standard libraries contain numerous functions. All other libraries are built with functions, too.

As you learn about various functions, you can always call documentation for help by using the `?` symbol before the function, like this:

```python
# what does round() do?
?round
```

Notice the `Signature: round(number, ndigits=None)` part of this documentation. That's telling you what __parameters__ the function can interact with. In this case, `round()` works with a `number` and `ndigits` (number of digits). These parametes work in that order.

So, to round a number using `round()` to the fourth digit, we would put in these parameters:

```python
# remember, Python is a zero-indexed language
# to round to the fourth decimal, we pass 3 to ndigits
round(3.14159, 3)
```

How would we round to the nearest whole number?

```python
round(3.14159, 0)
```

Function parameters are strict. Parameters need to be in the standard order. We also can't pass things to functions that don't work with them. For example, `round()` doesn't work like this:

```python
round(2, 3.14159)
```

The `TypeError` explains why: 'float' object cannot be interpreted as an integer. That's telling us that the data type we passed to the `ndigit` is not compatible.

We could just rearrange the parameters above to work properly. But also, you can specify the parameters to their __keywords__, if you like. That would look like this:

```python
round(ndigits=2, number=3.14159)
```

If you are so thorough as to specify your keywords, keep in mind that you must specify them all! Otherwise, the function won't run properly.

As you work with functions, keep using the `?` call for help. The more you read about them and interpret them, the easier it will be to remember how to use them. There are tons of functions in the Python standard library and the various libraries we'll be using. Take your time, keep looking them up and reading their documentation. It is a neverending process of remembering all the functions and their purposes.

### Writing Your Own Functions

That's right, you can create your own functions! In fact, you'll end up doing this often with advanced projects. Any time you come across a data task that will be repeated over and over again, it's best to design a function for that task.

This makes functions one of the most useful programming structures in Python. They're what makes Python so versatile and popular among DH research.

#### Basic Function Syntax

Writing simple functions is pretty easy. The first thing to note, however, is the __basic function syntax__. It looks like this:

![function syntax diagram](../img/functions.png)

This function converts feet to meters. It takes the parameter `feet` and multiplies it by .304. It saves the result as `meters` and returns `meters`. Here's what it looks like written out in Python:

```python
def feet_to_meters(feet):
    meters = feet * .304
    return meters
```

Notice how there is no output from running the block of code above. This is because defining a function doesn't run it. The function needs to be __called__, or run, with the appropriate parameters and arguments to execute the code it contains.

But now that we've defined `feet_to_meters()` in this session, we can use it:

```python
# feet to meters for 10 feet
feet_to_meters(10)
```

What if we pass a different data type to the function? What happens?

```python
feet_to_meters("ten")
```

We get a `TypeError`. "ten" as a string is not compatible with the code contained in our `feet_to_meters()` function. Be sure to keep your data types in mind when designing and using functions.

But if you want to incorporate `feet_to_meters()` in more detailed code, you can do that, too:

```python
total_feet = 10

total_meters = feet_to_meters(total_feet)

print("The feet to meters result is:", total_meters)

# Or like this:

total_feet = 200
print("The feet to meters result is:", feet_to_meters(total_feet))
```

#### Pseudocode

One common practice in writing functions and other complicated code is to first write out __pseudocode__. There's no standard format for pseudocode––it's just listing and defining what you want your functions to do before you try to translate those ideas into Python.

Writing pseudocode is a great practice as you begin to think and learn with Python. If you're good at writing pseudocode, your functions will be easier to understand and implement. Good pseudocode also makes working with LLMs to draft Python much easier and effective.

Let's consider a hypothetical scenario where we could use a function and we want to conceive of it in pseudocode before trying to build it in Python. Let's say we're trying to prepare some textual data. We know we'll need it to be lowercased. This ensures all uppercase and lowercase instances of words will be counted the same. And we know the text needs to be separated by commas. This will make it easier to put the text into lists at later steps.

We don't need to figure out the whole text analysis pipeline, but we know we need the following:

1) lowercase all letters
    - use .lower() method
2) separate words with commas
    - use .replace() method

This is some basic pseudocode. Let's try it by building a function applies it:

```python
# define the function and its parameter
def prepare_text(text):
    # string method .lower() to lowercase text
    text = text.lower()
    # .replace() method to replace spaces with commas
    text = text.replace(" ", ",")
    # always finish functions with return statement
    return text
```

Now let's try it on an example:

```python
example_text = "We're testing a function on some text!"

prepare_text(example_text)
```

It works! Now, moving forward, we can just run `prepare_text()` to make these lowercasing and comma separating changes in one shot.

What's more important, though, is thinking through coding steps and problems using __pseudocode__. Be sure to take your time and think things through before trying to write code. It will actually save you time, I promise!

One last thing––it's not required, but it is good practice to add `docstrings` to your functions. These create documentation that explains the function. It means you can use `?` to remind yourself what a function does. Of course, if you wrote the function, you'll probably remember what it does (unless it's been a while since you used it). But `docstrings` are still nice to have.

You can add them by putting text between three quotations `"""` at the beginning of the function, like this:

```python
def prepare_text(text):
    """
    Function to lowercase text and replace spaces with commas.
    """
    text = text.lower()
    text = text.replace(" ", ",")
    return text
```

And now, `prepare_text()` works with the `?` help call:

```python
?prepare_text
```

## Conditionals

Another fundamental structure in programming is the __conditional__. These blocks create certain rules for the given code. They will allow or prevent code to run depending on the conditions––on where specific things are true.

### If-Statements

The most widely used conditional is the __if-statement.__ An if-statement controls whether some block of code is executed or not.

- The first line opens with the `if` keyword and contains an expression to be evaluated. It ends with a colon.
- The body of the if-statement is indented. It contains the code to execute __if the condition is met.__ If it is not met, the lines of code will be skipped.

Let's look at an example:

```python
# define a variable
number = 100

# if-statement to check if number is greater than 50
if number > 50:
    print(number, "is greater than 50.")
```

As you can see, `number` set to 100 is greater than 50. Therefore, the if-statement prints the next line of code.

But what if the if-statement checks for a number greater than 100?

```python
# define a variable
number = 100

# if-statement to check if number is greater than 101
if number > 101:
    print(number, "is greater than 101.")
```

There is no output. The code doesn't print the statement unless the if-statement condition is met.

### Else-Statements

__Else-statements__ supplement if-statements. They allow us to specify an alternative block of code to run if the if-statement conditional evaluates to `False` (reads that the condition hasn't been met). 

Let's look at an example:

```python
# define a variable
number = 49

# if-statement to check if number is greater than 50
if number > 50:
    print(number, "is greater than 50.")
# else-statement to check if number is not greater than 50
else:
    print(number, "is not greater than 50.")
```

Else-statements are useful in cases when your data may have two possible qualities that you need to account for. Or, sometimes, you'll want to change your data, but only if it meets certain criteria, like this:

```python
# set the number variable
number = 40

# number must increase by 5, but only if it is less than 30
if number < 30:
    number = number + 5
    print(number)
# number must decrease by 5, but only if it is not less than 30
else:
    number = number - 5
    print(number)
```

### Elif-Statements

In many cases, we may want to check several conditionals at the same time. __Else-if (Elif-)__ statements allow us to specify as many conditional checks as we'd like in the same block.

Elif-statements must follow an if-statement. They only are checked if the if-statement fails. Then, each elif-statement is checked, with their corresponding bodies run when the conditional evaluates to `True`.

An else statement at the end can act as a "catch all", when the if statement and all following else-if statements fail.

In Python, else-if statements are indicated by the `elif` keyword. Here's an example:

```python
# set the temperature
temperature_fahrenheit = 78

if temperature_fahrenheit > 89:
    print("It's hot outside!")

elif temperature_fahrenheit == 78:
    print("It's perfect outside!")

elif temperature_fahrenheit > 69:
    print("It's warm outside!")

elif temperature_fahrenheit > 50:
    print("It's cool outside!")

else:
    print("It's cold outside!")
```

### Booleans

Remember that __logicals__ or __booleans__ are a data type. The if-statements we've just looked at are based on __booleans__. Booleans are variables that are __binary: they can either be `True` or `False`.

When we run if-statements, Python determines which block fo code should be executed based on the truth value of the condition. In other words, it reads and understands the data variables passed into the if-statements based on their truth value. Booleans therefore allow for decision making.

Remember: Booleans are also the result of so-called comparison operators, which are operators that compare two values. For example, equality is signaled in Python (and many other languages) by the double equals sign `==`. It's distinct from the assignment operator (single equals sign `=`) used in variable assignment.

## Iteration

Iteration gets at the heart of why we often turn to computers for DH work. It refers to any task that is repeated through computation. Computers are great at iteration. Once you define a task and identify data, iteration allows you to cycle through the task across data far faster than a human could do.

In Python, we often design iterative processes using __loops__. Loops execute code once for each value in an iterable data type or structure. Loops are written with `for` statements. They follow the pattern `for item in iterable_object:`.

Let's look at an example:

```python
# a list of average life expectancies
life_expectancy_list = [78.8, 82.3, 74.5, 80.1, 77.2]

# iterate through the list and round the life expectancies to the nearest whole number
for life_expectancy in life_expectancy_list:
    rounded_life_expectancy = round(life_expectancy)
    print(rounded_life_expectancy)
```

Notice how that reads? Translated to human language, it says:

"__For__ each life expectancy __in__ our list of life expectancies,

create a variable set to the rounded life expectancy,

and then print the rounded life expectancy."

Notice also how the for-loop defines the individual variables in `life_expectancy_list` as `life_expectancy`. This variable gets written over each time the loop runs over the next variable in the list.

What do you think `life_expectancy` is assigned to at the end of the loop?

```python
print(life_expectancy)
```

It remains the last entry of the list––the last data that was iterated over.

The loop also defined a new variable called `rounded_life_expectancy`. What is that variable after the loop is finished?

```python
print(rounded_life_expectancy)
```

### Loops with Strings

Loops can loop over any iterable data type or structure. Iterable data must have more than one datum and a sequence. Generally, anything that can be indexed (e.g. accessed with `values[i]`) is an iterable.

This is true for strings. Just as a reminder, how do we review the indexes of strings?

```python
example_string = "hasta la vista, baby"

print(example_string[0])
print(example_string[1])
print(example_string[2])
print(example_string[3])
print(example_string[4])
```

That's easy enough, but printing each character is an iterable task. Let's try to put it into a loop:

```python
example_string = "hasta la vista, baby"

for character in example_string:
    print(character)
```

Loops can get more intricate and complicated depending on what you want to include in them. For example, let's say we want to print the characters in `example_string`, update them uppercased text, and then print them again. We could do that like this:

```python
example_string = "hasta la vista, baby"

for character in example_string:
    print(character)
    character = character.replace(character, character.upper())
    print(character)
```

### Conditionals and Loops

But there's more. You can put conditionals––`if-else` statements––into loops!

Let's say we want to capitalize only the letter "a" when we print characters from `example_string`. And just for funsies, we want to print an exclamation point wherever there's a blank space. We could do those things by incorporating a conditional:

```python
example_string = "hasta la vista, baby"

for character in example_string:

    if character == "a":
        print(character.upper())

    elif character == " ":
        print("!")
        print(character)

    else:
        print(character)
```

It worked! Let's look at another example. This time, let's consider a list. Looping through lists with conditionals is a common practice––quite often, you'll have lists with some data that you want to do an operation on, and some data you don't. In these cases, loops and conditionals are your saving grace. 

Quite often, you'll want to extract those results, too. You can do that by creating a __shell__––a list or some other data structure to hold the parts of the list you want to save. This is also sometimes called an __accumulator variable__.

In this example, we're looking at the list `numbers`. We want to save all the numbers in our `numbers` list that are greater than 100. We'll put those greater numbers in a new list called `greater_numbers`. We have to define `greater_numbers` before we can use it in the conditional, though, so we'll assign it to an empty list––the empty `[]` brackets.

```python
# our list of numbers
numbers = [12, 20.2, 43, 88.88, 97, 100, 105, 110.9167]

# our empty list. We're defining it as an empty list so we can add to it later.
greater_numbers = []

# our for loop
for number in numbers:
    if number > 100:
        print(number, 'is greater than 100.')
        # here we add the number to our greater_numbers list
        greater_numbers = greater_numbers + [number]

print(greater_numbers)
```

Again, this is one of the most common usages for looping and conditionals. Let's consider one more example. Let's say we have another list of books (`book_list`). We want to create a new list of only the books in `book_list` that start with the capital letter "A".

In pseudocode, this might read like this:

- create shell to hold "A" letter books
- __for__ any book __in__ our book list:
    - use conditional: if
        - if the book starts with letter "A"
            - add the book to our shell

Can we turn that into Python?

```python
book_list = ["Beloved", "Anna Karenina", "The Great Gatsby", 
             "A Tale of Two Cities", "Invisible Man", "Pride and Prejudice", 
             "A Farewell to Arms"]

letter_a_books = []

for book in book_list:
    # if statement to check if book starts with "A"
    # uses the .startswith() string method
    if book.startswith("A"):
        # if the book starts with "A", add it to the letter_a_books list
        # uses the .append() list method
        letter_a_books.append(book)

print(letter_a_books)
```
