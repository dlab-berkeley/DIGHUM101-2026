### Working with Columns

There's more we can do with columns. Most pandas methods work with singular columns, too. We can look at just the first five rows of a column with `.head()`, for example:

```python
letters["writer"].head()
```

We can use `.describe()` on a column, too:

```python
letters["writer"].describe()
```

And again, `.value_counts()`:

```python
letters["writer"].value_counts()
```

You can also call multiple columns by passing their names in as strings to a double list! View just the first five rows:

```python
# That's a lot of square brackets!
letters[["writer", "date"]].head()
```

Why double brackets?

Single brackets `letters["writer"]` select a single column, and the result is a _Series_ –– not a list but a one-dimensional array which allows for different operations. This is just another data structure, one particular to pandas, but one we haven't covered.

Double brackets `letters[["writer", "date"]]` select multiple columns, and the result is a DataFrame.

There are many pandas methods that change or manipulate the dataframe. In terms of columns, one of them is `.rename()`. As it suggests, it's for renaming columns, like this:

```python
# capitalize column name
letters.rename(columns = {"writer" : "Writer"}, inplace = True)

letters.head()
```

Let's pause a moment. What does that `inplace = True` mean?

Try looking it up with `help()`:

```python
help(letters.rename)
```

What does inplace=True mean?

In pandas, many DataFrame methods (like rename, drop, sort_values, etc.) have an inplace parameter:

`inplace=True` This method modifies the original DataFrame directly and does not return a new DataFrame.

`inplace=False` (default) This method returns a new DataFrame with the changes, and leaves the original DataFrame unchanged.

This is a little safeguard built into pandas––it's meant to help you avoid overwriting your data on accident!

Also, did you notice we used the `{}` brackets? Those signal we used a dictionary. Let's say we wanted to update all the column names with capital letters. We could do that easily like this:

```python
letters = letters.rename(columns = {"writer" : "Writer", 
                          "source" : "Origin",
                          "destination" : "Dest",
                          "date" : "Date"})

letters.head()
```

```python
# notice the dictionary structure here
letters.rename(columns = {"writer" : "Writer", 
                          "source" : "Origin",
                          "destination" : "Dest",
                          "date" : "Date"}, 
               inplace = True)

# View the updated column names
print(letters.columns)

# or

letters.head()
```

You can also add or remove columns easily with pandas. If you wanted to add a column, you just do this:

```python
letters["Empty Column"] = ""
letters.head()
```

That created the column "Empty Column" with all empty values. To remove it, you just need to do this:

```python
letters = letters.drop(columns = ["Empty Column"])
letters.head()
```

Sometimes, you'll want to add columns based on certain values. For example, we've got the column "Date" already. Can we use it's values to create a column for "Year"?

Yes, yes we can. To do that, we'll just add a column like before, but assign its values to the first values from "Date" split at the "-" symbol, like this:

```python
letters["Year"] = letters["Date"].str.split("-").str[0]
letters.head()
```

That code also introduces `.str`––a pandas accessor that tells the code to read the column values as strings, as opposed to parts of a series. It's basically a shortcut in pandas for reading data types correctly.

Let's make columns for month and day, too:

```python
letters["Month"] = letters["Date"].str.split("-").str[1]
letters["Day"] = letters["Date"].str.split("-").str[2]
letters.head()
```

### Slicing Rows

You can slice rows like you would a string or a list. If we just want three rows:

```python
letters[60:70]
```

We can also look up individual rows using `.iloc`, like this:

```python
letters.iloc[10]
```

`.iloc` can also be used to return non-consecutive rows, if you pass it integers as a double list.

For example, to get the 4th, 12th, and 29th rows:

```python
# pesky Python zero index!
letters.iloc[[3, 11, 28]]
```

Then there's `.loc`, not to be confused with `.iloc`. They are very similar except for a key difference:

- `.iloc` looks up the rows by their integer positions
- `.loc` locates rows by their index labels

This is a subtle difference, but what it means is you can pass more than just integers into `.loc`. You can also pass column names, like this:

```python
letters.head(11)
```

```python
letters.loc[10, "Writer"]
```

There we've selected a single cell of data in our dataframe. You can select single cells with `iloc`, too, but you must translate them into integers (their coordinates in the dataframe):

```python
# the Writer column is the first column
# so it passes with 0
letters.iloc[10, 0]
```

### Conditional Subsetting

Sometimes, you'll want to narrow your dataframe based on certain conditions. For example, what if we just wanted a subset of data only when the Destination is equal to Haarlem?

... and the Writer is equal to Meulen, Andries van der?

In this case, we need to use conditionals. We can assess whether rows fit certain criteria by using conditionals like this:

```python
letters["Dest"] == "Haarlem"
```

```python
letters["Dest"] != "Haarlem"
```

We can also save these results using `.loc` and the conditional, like this:

```python
# Data frame just of Haarlem destinations...
haarlem = letters.loc[letters["Dest"] == "Haarlem"]
haarlem
```

This creates an entirely new dataframe:

```python
type(haarlem)
```

We could do this subsetting on author, too:

```python
# Data frame just of Andries van der Meulen
meulen = letters.loc[letters["Writer"] == "Meulen, Andries van der"]
meulen.describe()
```

And to bring it all together, let's write some pseudocode:

from letters, we need to __locate__ "Dest" values that are:

- equal to "Haarlem"
AND
- "Writer" values that are equal to "Meulen, Andries van der"

```python
# Data frame that includes both Haarlem as the destination AND Andries van der Meulen 
both = letters.loc[(letters["Dest"] == "Haarlem") &
                   (letters["Writer"] == "Meulen, Andries van der")]
print(both.shape)
both
```

It worked! What about any/all letters related to Haarlem?

In that case, we'd need:

from letters, we need to __locate__ Haarlem values in either:

- "Dest" column
OR
- "Origin" column

We can do that with the `|` symbol, like this:

```python
haarlem = letters.loc[(letters["Dest"] == "Haarlem") |
                     (letters["Origin"] == "Haarlem")]
print(haarlem.shape)
haarlem
```

If you want to learn more about pandas symbols, indexing, and so forth, check out the [documentation here.](https://pandas.pydata.org/pandas-docs/stable/user_guide/indexing.html)

### Sorting Values

Quite often, you'll need to sort the values of a (part of a) DataFrame. Pandas makes this easy with the `.sort_values()` method.

```python
letters.sort_values("Date", ascending=True)
```

```python
letters.dtypes
```

That sorts the whole column by the date values. Easy enough!

But sometimes, you'll want to sort narrower subsets. For instance, we can get only the letters with a destination of Delft, then sort the values in ascending order:

```python
letters[letters["Dest"] == "Delft"].sort_values("Date", ascending=True)
```

Let's break down all those brackets, though, because that's a little confusing!

Why do we need to structure this code like this?

`letters[letters["Dest"] == "Delft"]`

It's because this specifies the part of the dataframe we want sorted. It means:

- we want version of __letters__ where __letters__ column "Dest" is equal to "Delft"

With that in place, we then pass two parameters to the `.sort_values()` methods: the column "Date" and in what order (ascending)

`.sort_values("Date", ascending=True)`

There's another sorting method called `groupby()`. It "groups" rows together by their shared values.

```python
letters.groupby("Dest").describe()
```

Paired with other pandas methods, you can see how `groupby()` adds more computed information per value in "Dest". For every other column, we now have:

- count per value in "Dest"
- the number of unique values in the group
- the top value
- the frequency

In some cases, these can help you quickly compare values in a column as compared to their values in other columns.

## Pandas and Other Libraries

Pandas is highly compatible with all sorts of other libraries. This makes sense when you consider all the types of data that can be tabulated and kept in dataframes. Sometimes, you'll use other libraries to manipulate or compute data that is contained in dataframes.

Let's work with just one example: matplotlib. Matplotlib is a popular data visualization library. It seamlessly allows you to create visualizations of pandas dataframes.

First we'll import it with the alias `plt`:

```python
import matplotlib.pyplot as plt
```

Using Matplotlib, let's try to visualize one component of `letters`––the number of letters sent over the year.

To do that, we first need to isolate the part of the dataframe that will tell us this information. In our case, we'll look at the "Month" column. And using `.value_counts()` and `.sort_index()` to make sure our data is in the right monthly order:

```python
# Count how many letters appear in each month
letters_per_month = letters["Month"].value_counts().sort_index()
print(letters_per_month)
```

This is the part of the dataframe we want to visualize––the number of letters over time in months.

But what kind of data structure is it? What, exactly, are we visualizing?

```python
type(letters_per_month)
```

It's a series––a series representing the letters sent over time during the year in question.

In matplotlib, we can create a simple line plot with `.plot`.

We can also use `.title`, `.xlabel`, and `.ylabel` to create labels for our plot.

And then if we want to show the plot, we just run `plt.show()`:

```python
# Create a line plot
plt.plot(letters_per_month)

# Add labels
plt.title("Letters Over Time")
plt.xlabel("Month")
plt.ylabel("Number of Letters")

# Display the plot
plt.show()
```
