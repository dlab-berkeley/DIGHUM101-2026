### AI Policy

Large Language Models (LLMs) are creating dynamic, wide-ranging impacts for programmers and researchers. In just the past year, several frontier models have gained high capacity to generate functional code. We will be discussing these developments and LLMs in class. Using LLMs is not strictly prohibited, but there are three important considerations to bear in mind:

1) _Responsibility:_ The responsibility for all submitted work, whether text or code, lies with you. You should be ready to demonstrate your understanding of it and ensure that you have verified its accuracy.
2) _Honesty:_ Honesty regarding your use of these tools is of utmost importance. Submitting automatically generated code or text without appropriate citation will be regarded as plagiarism.
3) _Originality:_ While leveraging AI tools for assistance is permissible, the work you submit should predominantly reflect your original ideas, code, and language. Assignments that are in large part copied from an LLM are unlikely to be evaluated well.

In other words, there are permissible uses of AI in this class. It is acceptable to converse with LLMs about coding exercises or coding problems you're having trouble with. It is acceptable to use code generated by LLMs but you must cite your LLM usage and have a complete understanding of any code you submit. The majority of code you submit should also be written by you. Ignoring this policy would result in plagiarism, a serious offense that will be treated accordingly.

Most importantly, though, please remember that this class is meant to help you develop a working knowledge of Python in DH contexts. If you're relying on LLMs for every difficult part in the learning process, you need to pause and reflect. Why are you avoiding the challenging aspects of learning? What is it about the difficult parts that make you feel stuck? No matter how useful LLMs are or become, you'll never be able to code effectively unless you work through the difficult aspects of coding. Please be mindful about your learning processes and how LLM usage could be hindering them.

### 1) Anaconda

Anaconda is a free, open-source software suite used by data scientists, researchers, and programmers. You can think of it like Microsoft Office but for programming. It has lots of utility, but for our purposes, it's crucial to set up what's called a __Python environment__.

What's a Python environment? You can think of it like an isolated workspace on your computer, a specific directory that brings together these elements:

- a __Python interpreter__–the specific version of Python you're using. Python, like most anything run on computers, gets updated. Your Python interpreter is the version copy of Python that compiles and runs your code.
- the __standard library__-the essential Python functions that are built-in to the Python interpreter. You can think of the standard Python library as a core collection of tools in Python. It allows you to interact with files, do math, randomize variables–all sorts of stuff. 
- other __libraries__–we'll go over libraries in great detail throughout the class, but in a nutshell: they are collections of pre-written code intended to handle specific tasks. They are extensions of the standard library for specific use-cases (everything from scraping web data to machine learning). When you set up a Python environment, it's often helpful to choose the libraries you know you'll be using and download them into the environment from the start.

So, basically, we're using __Anaconda__ to create a __Python environment__ specifically designed for this class. To install Anaconda, follow these steps:

1) Go to [https://www.anaconda.com/download/success](https://www.anaconda.com/download/success)
2) Navigate to your device type (Mac or Windows)
3) Download __Miniconda__
    - You could download Anaconda Distribution instead, but it shouldn’t be necessary for this class. For more info about the differences, see here: [https://www.anaconda.com/docs/getting-started/concepts/anaconda-or-miniconda](https://www.anaconda.com/docs/getting-started/concepts/anaconda-or-miniconda)
    - If when you attempt to download Miniconda, you get an error message that reads something like "you can't download to this device", it probably means you're on an older device that, as of April 2026, is no longer compatible with the latest versions of Anaconda. Try going to [https://repo.anaconda.com/miniconda/](https://repo.anaconda.com/miniconda/). Then choose the file that matches your type of device and also ends with “.pkg”. This will allow you to download a past version of Anaconda suitable to your device.

Git is a __versional control system__–a tool that tracks, organizes, and manages changes to files. GitHub is an online platform extension of Git (a place online where Git commands can reach files). Git doesn't have a typical user interface. To use it, you must enter commands into the __Terminal__. The terminal is what a lot of people imagine when they think of coding. The Terminal may seem intimidating, but fear not: with Git, you only need to consider a specific selection of commands.
