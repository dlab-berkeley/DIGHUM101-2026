### 1) Floats

Floats are decimals. In Python, you need to differentiate between decimal values (things that can be split apart) and integer values (things that can't be divided past `1`).

```python
# calculate the volume of a box
# assume measurements are in inches
length = 5.5
height = 4.2
width = 3.1

volume = length * height * width

print("The volume of the box in inches is:", volume, "cubic inches.")
```

 Another way to think about floats is to think of them as more precise than integers. Consider, for example, having to render pi as an integer: 3. It becomes far less precise when calculating volume of a circle.

```python
# pi as an integer

radius = 2.5
height = 4.2

pi = 3
volume = pi * (radius ** 2) * height
print(volume)
```

```python
# pi as a float
radius = 2.5
height = 4.2

pi = 3.14159
volume = pi * (radius ** 2) * height
print(volume)
```

### 2) Integers

Integers are whole numbers (negative and positive), including zero such as -3, 0, 7, and 8. They're best used to count sequences of things.

```python
apples = 5
grapes = 3
grapefruits = 4
total_fruits = apples + grapes + grapefruits
print("There are", total_fruits, "fruits total.")
```

Floats and integers can be used together. When you do this, however, it's important to note that Python will treat the output as a float.

```python
apples = .5
grapes = 3.5
grapefruits = 4
total_fruits = apples + grapes + grapefruits
print("There are", total_fruits, "fruits total.")
```

If you're unsure what data type a number is, you can check with the `type()` function, like this:

```python
type(total_fruits)
```

If you want convert a whole number back to an int, or vice versa, you can reassign it with `int` or `float` respectively. 

```python
total_fruits = int(total_fruits)
print(total_fruits)
print(type(total_fruits))
```

```python
total_fruits = float(total_fruits)
print(total_fruits)
print(type(total_fruits))
```

### 3) Strings

Although there are some technical differences, we will use use "string" as synonymous with text/character data. String data should always be wrapped in quotations. Single or double quotes are both fine, but be consistent! You can nest single quotes inside of double quotes.

```python
"I'm a string!"
```

You can also use the backslash `\` (also called escape character) to signal the literal use of a character that is giving your string trouble, like this:

```python
'I\'m a string!'
```

The escape character has some other uses ([learn more here](https://www.w3schools.com/python/gloss_python_escape_characters.asp)), making it important when you're working with strings. Here's a brief example of all the ways you can delineate strings with single or double quotations:

```python
# each status variable is the same string
# but they differ in how they are written
status = "I am hungry"
status = 'I am hungry'
status = "I'm hungry"
status = 'I\'m hungry' 
print(status)
print(type(status))
```

#### Strings Have Methods

Methods are object-specific functions. You can look up methods for the `status` variable by typing `status.`. When you do that in VS Code, a menu of method options should appear. With strings, you can do things like `.upper()` or `.lower()` or `.capitalize`:

```python
print(status.upper())
print(status.lower())
print(status.capitalize())
```

With string methods, you can even count the number of specific words in a string:

```python
daft_punk_lyrics = "around the world around the world around the world"
daft_punk_lyrics.count("around the world")
```

Just an FYI: other data types and structures have methods, too. You can look them up by adding a period after the type, like this:

```python
float_example = 3.14159
float_example # add . to look up methods
```

```python
int_example = 50
int_example # add . to look up methods
```

### 4) Logical Statements

The logical type refers to True or False values. Logical data, also called booleans, are useful for many scenarios, such as turning function parameters on or off and logical testing.

In Python, booleans are treated like special numeric values when used in mathematical operations. `True` is treated as `1` and `False` is treated as `0`. This a funny fact about Python. It means that if you do the following equation:

```python
# What is going on here? 
False - True
```

You get -1 because Python effectively interpreted the equation as `0 - 1`. This behavior exists because Python’s boolean type is closely related to integers under the hood. It allows booleans to participate in calculations, counting operations, and logical tests.

There are also symbols used to refer to logicals. They include:

- `==`: checks whether two values are equal
- `!=`: checks whether two values are not equal
- `>`: checks whether one value is greater than another
- `<=`: checks whether one value is less than or equal to another

You can use these comparison operators to produce logical values (True or False). Python evaluates each comparison statement and determines whether the statement is logically correct, like this:

```python
print(7 == 7)
print(7 != 7)
print(8 > 7)
print(8 <= 7)
```

#### Indexing, Slicing, and Striding

##### Indexing

Indexing is a way to access just a part of some variable by referring to the position of that part. We can do this by using `[]` brackets––naming the variable and putting our intended index in the brackets. 

Consider this string, for example:

```python
pangram = "The quick brown fox jumps over the lazy dog"
print(pangram)
print(type(pangram))
```

If we just want the first letter of this sentence, we could just type the number 1 in square brackets to refer to the first letter `T`, right?

```python
pangram[1]
```

__Wrong!__

Turns out, Python is a "zero-indexed" language, meaning counting starts at zero. Therefore, we would need to type `[0]` to reach the first index. This is a famous quirk of Python. Perhaps infamous? Idk, you get used to it.

```python
pangram[0]
```

##### Slicing

Sometimes you'll want to index a range of values. This is possible with slicing––calling multiple indices by placing a colon inside of our square brackets `[:]`. The index before the colon is included in the slice, but the number after the colon is excluded because of Python's zero-indexing.

So, `[0:4]` will call indices 0, 1, 2, and 3:

```python
# Slice only the first four indices:
pangram[0:4]
```

Notice how this captured the space after "The"? Remember: we're dealing with strings which are technically not the same thing as text.

You can also call ranges of indices within strings, like this:

```python
# slice from the 5th to the 9th index:
pangram[4:9]
```

You can also slice strings backwards, like this:

```python
# Slice only the last three characters:
pangram[-3:]
```

##### Striding

Striding is a way to call string indexes with an interval or step. You can do this by including a colon to specify the interval, like this:

```python
# Slice 0:12, but by every second index
pangram[0:12:2]
```

The `2` after the colon sets the stride.

### 5) Lists

A list is a mutable, ordered, indexable collection of data. Each value is contained within square brackets and separated by a comma, like this: `[value_1, value_2, value_3]`.

Lists have a lot of utility. Imagine, for example, you're researching literature and you have a list of titles. You can store this information in a __list of strings.__

```python
library = ["Catch-22", "The Brothers Karamazov", "Little Women", "Crime and Punishment", 
           "The Alchemist", "Rashomon", "Catch-22", "Pride and Prejudice"]

library
```

Data structures, including lists, can also have their type verified with the `type()` function:

```python
type(library)
```

Lists can also be counted. To count the length of a list, you can use the `len()` function: 

```python
len(library)
```

#### Indexing and Slicing Lists

We can also index and slice lists just like we did with strings.

```python
print('The first book in library is:', library[0])
print('The fourth book in library is:', library[3])
print('The fourth and fifth books in library are:', library[3:5])
print('The first two books in library are:', library[:2])
print('The last two books in library are:', library[-2:])
```

#### Overwriting List Elements

We can overwrite individual elements in a list using variable assignment. Note that lists can contain values of different types!

```python
# Where did The Brothers Karamazov go?
library[1] = "Things Fall Apart"
library
```

```python
# What is going on in this assignment??
library[2], library[3] = 5, True
library
```

#### Lists Have Methods

Just like strings (or mostly anything else in Python), data structures have methods as well - specific functions that only apply to them. Type the name of your list followed by a period. A list of methods you can call should appear:

```python
library.
```

One useful list method is counting. It's as simply as this:

```python
# How many times does Catch-22 appear in library?
library.count("Catch-22")
```

Another useful list method is `.split()`. It replaces a separator (usually a space) by splitting a string into multiple pieces and storing them as items in a list. This is especially useful when breaking sentences or lines of text into individual words.

```python
book = "Go Tell It on the Mountain"
book.split()
```

You can also assign a character as the separator using `.split()`. For example, let's say you've got your groceries recorded as a string but you want to turn them into a list separated by commas. You can do that like this:

```python
groceries = "milk,eggs,bread,cheese"
groceries.split(",")
```

#### Deleting List Elements and Variables

Deleting elements of a list or an entire list can be done using `del`

```python
# Delete just one item in a list - the element 7 is gone:
print(library)
del library[2]
print(library)
```

Careful with `del` though!

You may accidentally delete the whole thing.

```python
# Delete the entire list
del library
print(library)
```

#### Lists of Lists

Yes, you can make lists of lists, and lists of more lists. So many lists!

```python
list_of_lists = [[6, 3, 5], 
                 ["mountain", "forest", "ocean"], 
                 [5.0, 4.25, 12.823], 
                 True, 
                 False
                ]

list_of_lists
```

And the lists of lists have indices, too. You can print the first list in the list of lists, and the third index of the second list in the list of lists:

```python
print(list_of_lists[0])
print(list_of_lists[1][2])
```

### 6) Dictionaries

A python dictionary is an ordered (in new versions of python) collection of key/value pairs. The key is the name of the data. The value is the data itself. By calling the key, you can always find its corresponding value. Keys must be unique and immutable. Strings and integers are commonly used as keys. The values can be anything, including lists, and even other dictionaries.

Each key/value pair is denoted by a colon `:` with the key on the left side and the value on the right. Each key/value pair within the dictionary is separate by a comma.

NOTE: we have seen round parentheses `()` for functions, square brackets `[]` for indexing/slicing, but now note the use of curly braces `{}` to define a dictionary. Picking up on these little details will help you start reading code more quickly.

```python
# example of a dictionary
poets_dictionary = {
            "name": "Forough Farrokhzad", 
            "year of birth": 1935, 
            "year of death": 1967, 
            "place of birth": "Iran", 
            "language": "Persian"}

print(poets_dictionary)
print(type(poets_dictionary))
```

#### Dictionary Methods

Just as before, let's check out the dictionary methods using the `.` approach:

```python
poets_dictionary.
```

One useful dictionary method is `.keys()`––it allows you to print just the keys of the dictionary to know what variables are kept in it.

```python
poets_dictionary.keys()
```

You can do the same to review just the values with `.values()`:

```python
poets_dictionary.values()
```

You can also call a particular key to see its value, like this:

```python
# What language did Forough speak?
poets_dictionary["language"]
```

#### Overwriting Dictionary Elements

Like lists, we can also overwrite values (whatever they might be) in place:

```python
poets_dictionary["language"] = "Farsi"
poets_dictionary
```

We can also add new key/value pairs:

```python
poets_dictionary["gender"] = "Female"
poets_dictionary
```

#### Nested Dictionaries

Dictionaries tend to have nested elements, meaning they have higher level keys with entries that contain keys and data for the higher level example.

Here's what a nested dictionary looks like:

```python
poets_dict ={
    "Forough Farrokhzad":{
        'name': 'Forough Farrokhzad',
        'year of birth': 1935,
        'year of death': 1967,
        'place of birth': 'Tehran',
        'language': 'Farsi',
        'gender': 'Female'
    },
    "Emily Dickinson":{
        'name': 'Emily Dickinson',
        'year of birth': 1830,
        'year of death': 1886,
        'place of birth': 'Amherst',
        'language': 'English',
        'gender': 'Female'
    }
}
poets_dict
```

To navigate these dictionaries, you can enter the first order key then the next order key:

```python
poets_dict["Emily Dickinson"]["language"]
```

You can also review the keys for all entries, but this moves us into another order of complexity, using __loops__.

```python
for poet in poets_dict.values():
    print(poet["name"], "wrote in", poet["language"])
```
