## Libraries

Libraries are additional pieces of software designed with––and usable with––Python. They're not like other software you might be familiar with, though. They don't come with user interfaces or anything like that. They're better understood as sets of pre-written Python code for specific tasks.

Your "base" Python installation comes with a lot of [built-in standard libraries.](https://docs.python.org/3/library/index.html) However, you might not want to reinvent the wheel everytime you do some specific analysis. That's where libraries come in. There are some very popular libraries that we'll be using in this class. They make our coding more efficient, easier to read, and easier to reproduce.

Think of a library as containing a bunch of shortcuts for you to use!

### Some Jargon

Many people use the terms modules, packages and libraries interchangeably, but they can be used to mean slightly different things:

- A __module__ is a bunch of related code saved in a file with the extension .py
- A __package__ is a collection of modules.
- A __library__ is an umbrella term for a reusable chunk of code. Usually, a Python library contains a collection of related modules and packages.

This is just a helpful distinction. It's also important because sometimes, you may not need an entire library––you just need a specific module or package that's part of the library.

To import specific modules or packages, you just have to structure the `import` code like this:

````python
from matplotlib import pyplot
````

`pyplot` is a module in the library `matplotlib`. In the above code, we're just importing the module, not the whole library.

#### Aliasing

Using an __alias__ for a module or library is a common practice in Python. There's no technical reason for it––aliases are just abbreviations people use so they don't have to type out `pandas` over and over again, or whatever the library might be.

Here's a common alias for `pyplot`, for example:

````python
from matplotlib import pyplot as plt
````

Notice the `plt` at the end? That's now the code session's term for `pyplot`. In other words, it is now the alias for the module. This alias is a shorthand way to call `pyplot` functions from the `matplotlib` library.

But what have we just imported into our session, you ask? What is `plt`?

We'll be discussing `matplotlib`, `pandas`, `nltk`, `numpy`––all these libraries and more over the next few weeks. So don't feel like you need to know everything about all of them right away. For now, just know that `matplotlib` is a popular library for creating visualizations.

You can read its documentation [here](https://matplotlib.org/stable/index.html). As you work with libraries, you'll end up reading their documentation, looking up use-cases and examples, asking LLMs to explain them, etc. As we've said, this is the most time-consuming aspect of coding! Don't shy away from looking up answers to your questions. All programmers need to look things up constantly.

Anyway, matplotlib has a lot of cool functions that can be used my typing `plt.`, like this:

````python
# plot a histogram
plt.hist([1, 2, 3, 4, 5, 6, 25])
````

Notice the `.hist` indicating histogram? This is calling the `matplotlib` histogram function. Since we're using the alias, it's a lot less typing. Otherwise we'd have to do this:

````python
# matplotlib full name
# semicolon removes the text output of the plot
matplotlib.pyplot.hist([1, 2, 3, 4, 5, 6, 25]);
````

What else can `matplotlib` do? LOTS. Here's another example, showing how you can use it to build visualizations line-by-line:

````python
# Randomly generate 7 x and 7 y values
x = [1, 2, 3, 4, 5, 6, 44]
y = [3, 2, 3, 1, 3, 2, 46]

# matplotlib has a lot of built in styles
# let's use the seaborn style for our scatterplot
# and see how it looks!
plt.style.use("seaborn-v0_8")

# Make a scatterplot with the .scatter() method! 
plt.scatter(x, y, color = "darkblue", marker = "x");
````

## Importing Data

So far, we've only looked at tiny example data like this:

```Python
x = [1, 2, 3, 4, 5, 6, 44]
y = [3, 2, 3, 1, 3, 2, 46]
```

These hand-keyed examples of data are great for instruction, but they're rarely how you'll interact with data when you code. Most of the time, you'll need to __import data__ from files on your computer or from files hosted online.

There are a number of libraries and modules that help to do this. Most importantly, though, is first knowing where your data is stored and where your working directory (the directory you're currently in).

To check the working directory, you just need to run `!pwd`:

````python
# stands for "print" working directory
!pwd
````

Because this notebook is stored in the week_1 directory, that's its default directory. If we had a data file in the same directory––for example, a copy of frankenstein.txt––we could import the data easily. But we don't. In this class, we're storing data in the `data/` directory.

To access the `data/` directory, we can use the `os` module. It's part of the standard Python library. It allows you to easily move between directories:

````python
# Import the os module
import os

# Another way to check the cwd ("current" working directory) using the os module
print(os.getcwd())

# use .. to move up one directory
os.chdir("../data")

print(os.getcwd())
````

The best way to think about moving through directories is to think of them as layers of a tree. In the case of our course materials, the base level of the tree is `DIGHUM101-2026`. That folder contains all other folders. `data` and `week_1` are branches of `DIGHUM101-2026`.

The more `../` you list to `os`, the more levels you'll move up in the process. If you wanted to go up one level from `data/`––to `DIGHUM101-2026`––you could just do this, though:

````python
print(os.getcwd())

os.chdir("..")

print(os.getcwd())
````

If you had a folder within a folder, you'd have to do `os.chdir("../..")` to go two levels up. And so on.

If you wanted to get back to the `data/` directory, you could then do this:

````python
# this works because we're currently in the DIGHUM101-2026 directory
# this tells os to change the directory to the data subdirectory
os.chdir("data")

print(os.getcwd())
````

Since we're here in `data/`, we maybe want to check what's in the directory. You can do that with `listdir()`:

````python
os.listdir()
````

Now that we're here, we can interact with the data. If we wanted to work with frankenstein.txt, for example, we just have to use `open()`: 

````python
frank = open("frankenstein.txt")
print(frank)
````

What does this mean?

`<_io.TextIOWrapper name='frankenstein.txt' mode='r' encoding='UTF-8'>`

It's telling you what frank is in quite literal computer language. That's because `open()` doesn't do anything other than create a connection to the file. But what if we wanted to read frank?

````python
# .read() actually pulls the text into Python
# careful! If the file is very large, this could cause problems
frank.read()
````

That's better, but not quite right. Perhaps Jupyter Notebooks aren't the place to read novels... You could, however, start interacting with frank by assigning it to an object, like this:

````python
frank = open("frankenstein.txt")

frank_text = frank.read()

# what kind of data type is frank_text?
type(frank_text)
````

If frank_text is a string, perhaps we can slice it?

````python
print(frank_text[:1000])

len(frank_text)
````

And here you can see how quickly we move from importing data to interacting with it. But we should pause a minute and point out some details here.

Firstly, in the above examples, we navigated to the data file. Sometimes, that's most convenient. Other times, however, you may rather point directly to the file. You can do that with `open()` as well, just as long as you have the full file path on your device. For me, that would look like this:

````python
# see the full file path?
frank = open("/Users/matthewkollmer/DIGHUM101-2026/data/frankenstein.txt")

frank_text = frank.read()

print(len(frank_text))

print(frank_text.count("London"))
````

To get the full file path for any file in VS Code, just right click the file and select `copy path`. This is usually the most convenient way to import data files. But sometimes, you'll want to change the working directory, too, so it's good to know how to do both.

Secondly, you may have noticed that frankenstein.txt was a `.txt` file. This file format is compatible with base functions `open()` and `read()`, but other file formats are not. As you work more in Python, pay attention to file formats. They will require different approaches and libraries.

Consider, for example, `.csv` files. They are quite common in DH work and data analysis. To import them, you need a library called `pandas`. This is probably the library you'll use the most in your DH work. It's a popular library for dealing with tabular data, especially `.csv` files.

Here's how to import `pandas` into your session:

````python
import pandas as pd
````

To use it to import a `.csv` file, you'll use its `read_csv()` function. Let's try it on the books_search.csv, just for funsies:

````python
# check working directory
print(os.getcwd())

books_search = pd.read_csv("books_search.csv")
````

That's it! Now we've got the books_search.csv loaded into our session as an object called `books_search`. What kind of object is `books_search`?

````python
type(books_search)
````

It's a pandas dataframe. A dataframe is the data type that pandas has converted the books_search.csv into. If you want to get a sense of what's in it, you can use the `.head()` function to read the top rows:

````python
books_search.head()
````
