## Working with Jupyter Notebooks

We use Jupyter as our interface for coding. The document you're looking at right now is called a __Jupyter Notebook__. You can think of it like a hybrid Microsoft Word document or Google Doc. It allows you to text, but also executable code. It also lets you render images and visualizations. In the coding world, Jupyter Notebooks are popular because they make code far easier to interpret.

### Notebook vs Script

In this class, we'll mostly use Jupyter Notebooks, which let you write code in small sections (called cells) and see the results right away.

But there's another way to write Python code: scripts.

A script is just a plain text file (usually ending in .py) that contains a list of instructions written in Python. Instead of running one small piece at a time (like a cell in a notebook), you usually run the whole script from beginning to end––like pressing "Play" on a movie.

Scripts are often used in more advanced or professional settings when:

- you want your code to run all at once
- you don't need to explain each step
- you're building a program to run repeatedly or automatically

Scripts, in other words, are a way of coding that is more commonplace when the systems just need to run without the user needing to understand or interact with every step. Notebooks, on the other hand, allow users to read human explanations of things and understand each part of the code.

We won't be using scripts in this class, but it's helpful to know what they are and when they might be used.

## Command Mode and Edit Mode

Jupyter has two modes: Edit mode and Command mode. Edit mode allows you to type into the cells like a normal text editor. Command mode allows you to edit the notebook as a whole, but not type into individual cells.

Enter __Edit Mode__ by pressing __Return__ on Mac, or __Enter__ on PC. You can also double-click on a cell’s editor area.

Enter __Command Mode__ by pressing __Esc__ or clicking outside a cell's editor area. In command mode you can use shortcuts to work with cells:

- `c`: copy cell.
- `v`: paste cell.
- `d, d` (press key twice): delete cell.
- `a`: insert cell above.
- `b`: insert cell below.

You can also use the controls at the top of the Jupyter window––the "__+ Code__" or "__+ Markdown__" buttons, the __Run button__, the __checkmark button__––all these function fairly intuitively. 

## Two Types of Cells

In Jupyter Notebooks, code and text are divided into cells which can each be run separately. There are two types of cells: __Markdown cells__ and __code cells.__ You'll need to use them both to create fully drafted notebooks.

### Markdown

This cell is written in __Markdown,__ a lightweight markup language. Markdown allows you to format text with headings, bulleted lists, URLs, bold text, italics, and so forth. If you usually work in a text editor like Microsoft Word or Google Docs, you'll understand how to edit text with a visual interface. Markdown, on the other hand, eliminates the interface, meaning you just type certain keys to format text as you prefer.

For example, if you wanted to __bold some text__ in Markdown, you can either frame the text with two underscores or asterisks, like this:

`__bold text__` or `**bold text**`

Outside those backward apostrophes (called backticks), the underscores and asterisks will render the text like this:

__bold text__ or **bold text**

The backticks are part of Markdown, too. They they format the text as code regardless of the other symbols around the text.

To italicize text, you can frame the text with one underscore or asterisk, like this:

`_italicized text_` or `*italicized text*`

Here it is without the backticks:

_italicized text_ or *italicized text*

For headings, you use the # symbol. For every subheading, you just add a # symbol, like this:

`# Heading One`

`## Heading Two`

`### Heading Three`

`#### Heading Four`

`##### Heading Five`

And so on. Here's what they look like without the backticks:

# Heading One
## Heading Two
### Heading Three
#### Heading Four
##### Heading Five

If you want to hyperlink text, you put brackets around the text followed by parentheses containing the hyperlink. That creates a hyperlink [like this.](https://www.youtube.com/watch?v=dQw4w9WgXcQ)

Here's what it looks like when you're writing in a Markdown cell:

`[text goes here](full URL goes here)`

Here's a [cheatsheet](https://github.com/adam-p/markdown-here/wiki/Markdown-Cheatsheet) if you want to know more Markdown options. There are many. Once you get the hang of it, it's actually faster than writing in Microsoft Word. No need to use the mouse while drafting text!

In Jupyter Notebooks, you can double-click on any of the Markdown cells to see how they are made.

Try double clicking this Markdown cell to see what it looks like as Markdown code. Press __Shift + Return__ on Mac, or __shift + Enter__ on PC, to go back to the formatted text.

## Variables in Python

Perhaps the best way to introduce variables in Python is to point out that Python can also be used as a calculator. You can add:

```python
10 + 4
```

You can subtract:

```python
10 - 4
```

You can multiply:

```python
10 * 4
```

You can divide:

```python
10 / 4
```

And Python follows the usual PEMDAS order of operations:

```python
(10 + 4) * 2
```

This is cool, but how do we save the results of our calculations?

In Python, we can assign a value to a __variable,__ using the equals sign `=`. You can think of a variable as a "placeholder" or symbolic name for some value.

For example, we can assign the result of the above calculation to a variable called `a`.

```python
a = (10 + 4) * 2
print(a)
```

From now on, whenever you refer to `a`, Python will substitute the value we assigned to it.

Tip: In Jupyter, you don't always need to use `print` explicitly. When you want to check what value a certain variable holds, you can just type the variable name and run the cell, like this:

```python
a
```

### Calculating with Variables

The key feature of variables is that we can use them to clarify and standardize components of a script or program.

For example, let's say you wanted to calculate the increase of a population over time. How could you do it with variables?

You would first need to define your variables. You'd need the `start_year` and the `end_year`. You'd need the `annual_increase`. You'd need the `starting_population`. Let's start there:

```python
# variables to calculate population growth
start_year = 2026
end_year = 2030
annual_increase = 500
starting_population = 5000
```

Now how do we use these variables to calculate total population increase? First we need to calculate the number of years we're tracking:

```python
# create years variable to calculate total population growth
years = end_year - start_year
years
```

Then we need to multiply our annual population increases by the number of years we're tracking:

```python
# calculate total increase
total_increase = years * annual_increase
total_increase
```

Then we just need to add the `total_increase` by the `starting_population`:

```python
# calculate final population
final_population = starting_population + total_increase
final_population
```

You might have noticed the `#` symbols in the code cells above. These are called __comments__ in the code. By adding the `#` symbol at the start of the line, you're telling the code cell it's not Python, just a comment for human readers.

You may also notice that __lines of code run sequentially__ in Python. This makes it easier to understand the order of operations for a given script. It also means the order of operations is fundamental to Python––and coding in general. If you put our above program together in one code cell, for example, it works just fine:

```python
# variables to calculate population growth
start_year = 2026
end_year = 2030
annual_increase = 500
starting_population = 5000

# create years variable to calculate total population growth
years = end_year - start_year

# calculate total increase
total_increase = years * annual_increase

# calculate final population
final_population = starting_population + total_increase
final_population
```

But if you put things out of order, the code won't run properly. Here's the same process with new variable names out of order. See how it doesn't run?

```python
# variables to calculate population growth (broken code)
start_year_broken = 2026
end_year_broken = 2030
annual_increase_broken = 500
starting_population_broken = 5000

# calculate total increase (broken code)
# notice how this line is out of place?
total_increase_broken = years_broken * annual_increase_broken

# create years variable to calculate total population growth (broken code)
years_broken = end_year_broken - start_year_broken

# calculate final population (broken code)
final_population_broken = starting_population_broken + total_increase_broken
final_population_broken
```

In this case, we got a `NameError`. `years_broken` is not defined. That's why it's important to keep your code in the proper order. As you work in Python more and more, you'll find this strict sequential thinking becomes more habitual and intuitive.

One other note on the above examples––you may have noticed that I changed the variable names between the functional and broken examples. This is because __variables remain defined throughout a notebook until you restart the Kernel or you re-run or redefine them in code cells__. In other words, variables stay the same during a coding session until you change them somehow.

Do you remember our first variable in this session? It was `a`. Is it still the same?

```python
a
```

That can be a tricky thing about Jupyter Notebooks. Sometimes, you'll have a variable you defined many steps prior that you've forgotten about. As much as possible, try to keep track of your variables!

### Naming Variables

There are some other simple rules about variables. They are:

- Variables cannot start with a digit
- Variables cannot contain spaces, quotation marks, or other punctuation
- Variables _may_ contain an undescore
- Variables are case-sensitive (`life_expectancy` is not the same as `Life_Expectancy`)

Ignoring these rules will result in an error in Python.

## Debugging

In the above example, we saw our first error message: `NameError`. This error means the variable, function, or module you're calling hasn't been defined in the code. There are many other errors. __Don't be daunted by them!__ They are inevitable parts of coding. In fact, when you're programming, most of your time will be spent debugging, looking stuff up, reading documentation, troubleshooting, testing. Comparatively little time is actually spent typing out the code. Even the most experienced programmers end up with lots of errors as they're trying to write new code. It's just part of the process.

In 1-3_errors-help.ipynb, we'll go over all the common error messages. For now, just know that you should approach errors with these guidelines:

2) __Check your syntax__: You might just be spelling something wrong or missing some necessary punctuation. This is quite common.
3) __Look for help__: You might just be using a function in a wrong way. Get into the habit of reading documentation and finding help online. We'll be doing this in the next workshops.

## The Kernel

This notebook also made mention of the __kernel__. The kernel is the computational engine that executes the code contained in a Jupyter Notebook. Each time you run a code cell, the kernel processes that cell, executes the code, and keeps a record of what was run.

The kernel, in other words, is the system that supports and keeps track of the executed code in a coding session. This is how Jupyter remembers all the lines of code it executed, __even if it's not currently displayed in the Notebook__. Deleting a line of code or changing it to Markdown does not delete it from the Notebook's memory if it has already been run. This can cause a lot of confusion.

### Restarting the Kernel

To clear your session in a Jupyter Notebook, use `Kernel -> Restart` in the menu. If you reset the kernel, it's as if you just opened up the Notebook for the first time. None of your code cells will be run and all your execution counts will reset. Also, all your variables will be erased until you define them again in the code cells.
