## Errors

This notebook will help you begin to understand the frustration of error messages, their meaning, how to fix them, and how to find help. Remember: every programmer encounters errors––no matter how novice or advanced you are. Once you know why you get certain types of errors, they become much easier to fix. Errors appear in what is called a __traceback__, or the series of functions showing why the error occurred. Let's focus on a few of the most common ones:

- `SyntaxError`
- `IndentationError`
- `NameError`
- `IndexError`
- `TypeError`
- `KeyError`
- `FileError`

### Syntax Error

A syntax error means there is a problem with the structure of your code. It means some necessary part is either missing or out of place.

Can you see what's wrong the code below? Always read the last line of the traceback first, and then work backwards.

```python
print("Typing mistakes are inevitable."
```

### Indentation Error

Another type of syntax error is an indentation error. We haven't touched on indentations yet, but they're an essential part of writing code. They signal a grouping of parts based on their level of indentations.

We'll go over indentations in greater detail soon. For now, just know a proper indent should consist of a single TAB or four spaces. Otherwise, you'll get an indentation error.

```python
def a_little_function():
    message = "This is a little baby function that prints a message."
  return message
```

### Name Error

A name error occurs when you try to use a variable that does not exist. This is usually because you forgot to define it or included a typo in its spelling.

```python
print(my_name)
```

### Index Error

Index errors occur when you try to reference something that has a sequential structure. Many data structures, such as strings, lists, dictionaries, or dataframes follow sequences. If you get an index error, it means the thing in the sequence you're trying to reference doesn't exist or is out of range. 

```python
# grocery list
groceries = ["sushi", "chicken", "pizza", "salad", "soda"]

# This works...
groceries[4]
```

```python
# But this does not work because there is no fifth index in groceries
# Remember: the fifth index references the sixth position, but we only have five elements in groceries
groceries[5]
```

### Type Error

Type errors occur when you confuse types, such as when we try to use a method or syntax on a type it is not intended for. You can't mathematically add a string and a integer, for example.

```python
"2" + 2
```

### Key Error

Key errors occur when you try to access elements of a dictionary using a key that does not exist.

```python
temperature = {"day":["Mon", "Tue", "Wed"],
              "temp":[88, 99, 65]}
temperature
```

```python
# notice there is no "month" key in the temperature dictionary
temperature["month"]
```

### File Error

File errors occur when we try to load data from files but the file doesn't exist or we are looking in the wrong directory.

## Getting Help

Python documentation is essential! It tells you everything you need to know once you learn how to navigate it.

Python documentation is free online: [https://docs.python.org/3/](https://docs.python.org/3/). You can also type a question mark after a function to call its help files.

```python
float?
```

As you can see, this provides a brief explanation of the function.

Let's try it for `int`, `str`, and `print`.

```python
int?
```

```python
str?
```

```python
print?
```

Once you load libraries, you can also use the `?` help call to understand the library's components. Take a look at the `.read_csv` function in Pandas as an example:

```python
import pandas as pd
pd.read_csv?
```

This provides a lot more info. The truncated version is a little confusing, but if you click `scrollable element` or open in a `text editor`, you'll be able to view the more comprehensive descriptions of `pd.read_csv()`.

### LLMs as Assistants

Large Language Models (LLMs)––especially frontier models––are especially useful when you're stuck on a common programming error—–whether it's a syntax mistake, a confusing traceback, or unexpected behavior in your code. Instead of scanning through multiple forum posts, you can paste your error message directly into the chat and ask for help. For example, if you’re using Pandas and see a KeyError, an LLM can often explain what it means, why it’s happening, and how to fix it. 

This is an acceptable use of LLMs in this class. They will often make code suggestions, too. You may consider LLM suggestions, but be sure to know what the code is doing line-by-line. Never just copy and paste an LLM output without understanding it completely! And always cite LLMs when you collaborate with them to create code.
