## Step 1: Load The Data

Run the next cell to load the IMDb subset directly from UCI. You do not need to edit this cell.

The file is tab-separated, with one review sentence and one sentiment label per row. The `quoting=csv.QUOTE_NONE` setting keeps pandas from treating quotation marks inside review sentences as CSV syntax.

```python
dataset_url = "https://archive.ics.uci.edu/static/public/331/sentiment%2Blabelled%2Bsentences.zip"
file_inside_zip = "sentiment labelled sentences/imdb_labelled.txt"

with urlopen(dataset_url) as response:
    zip_data = BytesIO(response.read())

with ZipFile(zip_data) as zip_file:
    with zip_file.open(file_inside_zip) as file:
        movie_reviews = pd.read_csv(
            file,
            sep="	",
            names=["review", "sentiment"],
            quoting=csv.QUOTE_NONE,
        )

movie_reviews.shape
```
